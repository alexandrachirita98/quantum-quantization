# VAE Paper Figures (Kingma & Welling, 2013) on Kaggle

Clones `quantum-quantization`, installs dependencies from `requirements.txt`, lets you pick a baseline VAE model, and plots Figures 2-5 from [Auto-Encoding Variational Bayes](https://arxiv.org/pdf/1312.6114) using `VAEPaperFigureHandler`.

## 1. Clone the repository

```python
import os

REPO_URL = "https://github.com/alexandrachirita98/quantum-quantization"
REPO_DIR = "/kaggle/working/quantum-quantization"

if not os.path.isdir(REPO_DIR):
    !git clone {REPO_URL} {REPO_DIR}
else:
    print(f"Repo already present at {REPO_DIR}, pulling latest changes.")
    !git -C {REPO_DIR} pull

SURVEY_ROOT = os.path.join(REPO_DIR, "survey", "src", "related_work", "vae")
print("SURVEY_ROOT:", SURVEY_ROOT)
```

## 2. Install dependencies from requirements.txt

If you re-run this cell after it already succeeded once, restart the kernel first — `--force-reinstall` on packages already imported into the running kernel (e.g. numpy/scipy) won't take effect until re-import.

```python
REQUIREMENTS_PATH = os.path.join(REPO_DIR, "requirements.txt")
# Kaggle images ship with a preinstalled numpy/scipy pair; a plain
# `pip install -r` can upgrade one but leave the other's compiled
# extensions mismatched (ImportError from scipy._core.umath). Force a
# consistent reinstall of the whole pinned set instead.
!pip install -q --force-reinstall --no-cache-dir -r {REQUIREMENTS_PATH}
```

## 3. Choose the baseline model

Set `MODEL_NAME` to `"novicestone"` (NoviceStone/VAE-style model, explicit `latent_size`/`hidden_size` constructor args) or `"pytorch"` (the standard `pytorch/examples` VAE, hardcoded 784-400-20-400-784 and rebuilt per-figure by the handler).

```python
MODEL_NAME = "pytorch"  # "novicestone" or "pytorch"
assert MODEL_NAME in ("novicestone", "pytorch")
```

## 4. Instantiate the figure handler

```python
import sys

sys.path.append(os.path.join(SURVEY_ROOT, "handlers"))
from vae_paper_figures import VAEPaperFigureHandler

import torch
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

handler = VAEPaperFigureHandler(model_name=MODEL_NAME, device=device)
```

## 5. Plot Figures 2-5

```python
import matplotlib.pyplot as plt

fig2 = handler.plot_figure2()
plt.show()
```
